```python
# import libraries
import numpy as np
import torch
```

# Using numpy

```python
# create a vector
nv1 = np.array([1,2,3,4])
nv2 = np.array([0,1,0,-1])

# dot product via function
print(np.dot(nv1,nv2))

# dot product via computation
print(np.sum( nv1*nv2 ))
```

# Using pytorch

```python
# create a vector
tv1 = torch.tensor([1,2,3,4])
tv2 = torch.tensor([0,1,0,-1])

# dot product via function
print(torch.dot(tv1,tv2))

# dot product via computation
print(torch.sum( tv1*tv2 ))
```

```python
#Dot product represents commonality between two vectors/matrices (like images, corelation/covariance between data)
#Practice of dot product with matrices

tm1 = torch.tensor([[1,2,3,4],[1,2,3,4]])
tm2 = torch.tensor([[0,1,0,-1],[0,1,0,-1]])

#print(torch.dot(tm1,tm2)) #Error torch dot allows only 1D dot product
print(torch.sum(tm1*tm2))


nm1 = np.array([[1,2,3,4],[1,2,3,4]])
nm2 = np.array([[0,1,0,-1],[0,1,0,-1]])

#print(np.dot(nm1,nm2)) #Error numpy expects inner dimensions to be equal
print(np.sum(nm1*nm2))

#Use matmul function for torch and numpy to multiply
print(torch.matmul(tm1.T,tm2))
print(np.matmul(nm1.T,nm2))
```
