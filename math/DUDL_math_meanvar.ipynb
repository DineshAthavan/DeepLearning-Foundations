```python
# import libraries
import numpy as np

#Mean gives the central data/average. Works well for normal distribution but not for non linear distribution.
#Variance gives the measure of how far the data is distributed from each other. It works on all distribution.
#Mean and variance are used for normalization. (L1 and L2 norms are calculated)
```

```python
# create a list of numbers to compute the mean and variance of
x = [1,2,4,6,5,4,0]
n = len(x)

# compute the mean
mean1 = np.mean(x)
mean2 = np.sum(x) / n

# print them
print(mean1)
print(mean2)
```

```python
# variance

var1 = np.var(x)
var2 = (1/(n-1)) * np.sum( (x-mean1)**2 )#Formula for variance

#Without squaring in the formula, sum would become 0. Abs could also be taken but squaring is more helpful since it is closer to euclidean distance and also extends
#naturally to concepts like root mean square.

print(var1)
print(var2)
```

```python
# uh oh!

var3 = np.var(x,ddof=1)# decrease degrees of freedom(ddof) represents how much value is subtracted from n in denomoinator. Default is 0.
#ddof of 0 means biased variance and 1 means unbiased(Refer statistical concepts for more info)
print(var3)
print(var2)
```

```python
# does it matter for large N?
#Difference between biased and unbiased results gets bigger as number of data points increases
N = 10000
x = np.random.randint(0,high=20,size=N)

var0 = np.var(x,ddof=0) # default
var1 = np.var(x,ddof=1) # unbiased

print(var0)
print(var1)
```

```python
import torch

x = [1,2,4,6,5,4,0]

x_tensor = torch.tensor(x,dtype=torch.float32)

print(x_tensor),print(' ')

mean1_tensor = torch.mean(x_tensor)
mean2_tensor = torch.sum(x_tensor) / torch.tensor(len(x_tensor),dtype=torch.float32)

print(mean1_tensor)
print(mean2_tensor),print(' ')

var1_tensor = torch.var(x_tensor,correction=0)#Biased
var2_tensor = torch.var(x_tensor,unbiased=True)#Unbiased(by default)

print(var1_tensor)
print(var2_tensor),print(' ')

#Calculating both with an inbuilt function

var3,mean3 = torch.var_mean(x_tensor)#unbiased(by default)
var4,mean4 = torch.var_mean(x_tensor,correction=0)#biased

print(mean3)
print(mean4),print(' ')

print(var3)
print(var4)
```
