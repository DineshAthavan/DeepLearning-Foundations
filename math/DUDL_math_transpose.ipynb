```python
# import libraries
import numpy as np
import torch
```

# Using numpy

```python
# create a vector
nv = np.array([ [1,2,3,4] ])
print(nv), print(' ')

# transpose it
print(nv.T), print(' ')

# transpose the transpose!
nvT = nv.T
print(nvT.T)

#Another way to transpose
nv = np.array([ [1,2,3,4] ])
print(np.transpose(nv))
```

```python
# repeat for a matrix
nM = np.array([ [1,2,3,4],
                [5,6,7,8]
               ])
print(nM), print(' ')

# transpose it
print(nM.T), print(' ')

# transpose the transpose!
nMT = nM.T
print(nMT.T)
```

# Using pytorch

```python
# create a vector
tv = torch.tensor([ [1,2,3,4] ])
print(tv), print(' ')

# transpose it
print(tv.T), print(' ')

# transpose the transpose!
tvT = tv.T
print(tvT.T)
```

```python
# repeat for a matrix
tM = torch.tensor([ [1,2,3,4],[5,6,7,8] ])
print(tM), print(' ')

# transpose it
print(tM.T), print(' ')

# transpose the transpose!
tMT = tM.T
print(tMT.T)
```

```python
#Practice

torch_row_vector = torch.tensor([[5,6,7,8]])
torch_col_vector = torch.tensor([[5],[6],[7],[8]])

numpy_row_vector = np.array([[5,6,7,8]])
numpy_col_vector = np.array([[5],[6],[7],[8]])

print(torch_row_vector)
print(torch_col_vector),print(' ')

print(numpy_row_vector)
print(numpy_col_vector),print(' ')

torch_col2 = torch.tensor([[1],[2],[3],[4]])
torch_matrix1 = torch.cat([torch_col_vector,torch_col2],dim=1)
torch_matrix2 = torch.cat([torch_col_vector,torch_col2])
print(torch_matrix1)
print(torch_matrix2)
```
